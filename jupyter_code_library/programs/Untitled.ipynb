```python
ASSIGNMENT 1 (ML)

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler

df=pd.read_csv("heart.csv")

# A. Cleaning
df.drop_duplicates(inplace=True)

# B. Missing Data
df.fillna(df.mean(numeric_only=True),inplace=True)
df.fillna(df.mode().iloc[0],inplace=True)

# C. Transformation
df=pd.get_dummies(df,drop_first=True,dtype=int)

# D. 75% Train, 25% Test
train,test=train_test_split(df,test_size=.25,random_state=42)

sc=MinMaxScaler()
train[train.columns]=sc.fit_transform(train)
test[test.columns]=sc.transform(test)

print("Training data:")
print(train.head())

print("\nTesting data:")
print(test.head())

print("\nTraining size:",train.shape)
print("Testing size:",test.shape)
```

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score,confusion_matrix

def evaluate_model(TP,FP,FN,TN):

    y_true=np.array([1]*TP+[0]*FP+[1]*FN+[0]*TN)
    y_pred=np.array([1]*TP+[1]*FP+[0]*FN+[0]*TN)

cm=confusion_matrix(y_true,y_pred)
accuracy=accuracy_score(y_true,y_pred)
precision=precision_score(y_true,y_pred)
recall=recall_score(y_true,y_pred)
f1=f1_score(y_true,y_pred)

print("Confusion Matrix:\n",cm)
print(f"Accuracy:{accuracy:.2f}")
print(f"Precision:{precision:.2f}")
print(f"Recall:{recall:.2f}")
print(f"F1 Score:{f1:.2f}")

sns.heatmap(cm,annot=True,fmt="d",cmap="Blues",
            xticklabels=["Pred Neg","Pred Pos"],
            yticklabels=["Actual Neg","Actual Pos"])
plt.title("Confusion Matrix")
plt.show()

evaluate_model(45, 55, 5, 395)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier,plot_tree
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report

df=pd.read_csv(r"C:\Users\Admin\Downloads\Admission_Predict (1).csv")
df.columns=df.columns.str.strip()

print("Original Dataset:")
print(df.head())

X=df.drop(["Serial No.","Chance of Admit"],axis=1)
y=(df["Chance of Admit"]>=0.5).astype(int)

features=X.columns

scaler=MinMaxScaler()
X=scaler.fit_transform(X)

Xtrain,Xtest,ytrain,ytest=train_test_split(
    X,y,test_size=0.25,random_state=42
)

model=DecisionTreeClassifier(max_depth=4,random_state=42)
model.fit(Xtrain,ytrain)

pred=model.predict(Xtest)

print("\nAccuracy:",accuracy_score(ytest,pred))
print("\nConfusion Matrix:")
print(confusion_matrix(ytest,pred))
print("\nClassification Report:")
print(classification_report(ytest,pred))

print("\nFeature Importance:")
for name,value in zip(features,model.feature_importances_):
    print(name,round(value,3))

plt.figure(figsize=(18,8))
plot_tree(model,feature_names=features,
          class_names=["Not Admitted","Admitted"],
          filled=True)
plt.show()
```

```python


import pandas as pd
from sklearn.model_selection import train_test_split,cross_val_score,GridSearchCV
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB,BernoulliNB

# a) Preprocessing
data=pd.read_csv(
 r"C:\Users\Admin\Downloads\SMSSpamCollection",
 sep="\t",names=["label","message"])

data["label"]=data["label"].map({"ham":0,"spam":1})

cv=CountVectorizer()
x=cv.fit_transform(data["message"])
y=data["label"]

# b) Train-Test Split
xtrain,xtest,ytrain,ytest=train_test_split(
 x,y,test_size=0.2,random_state=42)

# c) Two ML Algorithms
m1=MultinomialNB().fit(xtrain,ytrain)
m2=BernoulliNB().fit(xtrain,ytrain)

print("MNB Accuracy:",m1.score(xtest,ytest))
print("BNB Accuracy:",m2.score(xtest,ytest))

# d) Cross Validation
print("MNB CV:",cross_val_score(m1,x,y,cv=5).mean())
print("BNB CV:",cross_val_score(m2,x,y,cv=5).mean())

# e) Hyperparameter Tuning
p={"alpha":[0.1,0.5,1]}

g1=GridSearchCV(m1,p,cv=5).fit(xtrain,ytrain)
g2=GridSearchCV(m2,p,cv=5).fit(xtrain,ytrain)

print("Tuned MNB:",g1.score(xtest,ytest))
print("Tuned BNB:",g2.score(xtest,ytest))
print("Best MNB:",g1.best_params_)
print("Best BNB:",g2.best_params_)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score

# Load Dataset
df = pd.read_csv(r"C:\Users\Admin\Downloads\Mall_Customers.csv")
print("Original Dataset")
print(df.head())

# Data Preprocessing
df = df.drop_duplicates()
df = df.dropna()

le = LabelEncoder()
df["Genre"] = le.fit_transform(df["Genre"])

X = df.iloc[:, [3, 4]]
scaler = StandardScaler()
X = scaler.fit_transform(X)

# Train-Test Split
X_train, X_test = train_test_split(X, test_size=0.25, random_state=42)

# K-Means Clustering
kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)
kmeans.fit(X_train)
kmeans_labels = kmeans.labels_

print("K-Means Silhouette Score:", silhouette_score(X_train, kmeans_labels))

# Hierarchical Clustering
hc = AgglomerativeClustering(n_clusters=5)
hc.fit(X_train)
hc_labels = hc.labels_

print("Hierarchical Silhouette Score:", silhouette_score(X_train, hc_labels))

# Visualization
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.scatter(X_train[:, 0], X_train[:, 1], c=kmeans_labels)
plt.title("KMeans Clustering")
plt.xlabel("Annual Income")
plt.ylabel("Spending Score")

plt.subplot(1, 2, 2)
plt.scatter(X_train[:, 0], X_train[:, 1], c=hc_labels)
plt.title("Hierarchical Clustering")
plt.xlabel("Annual Income")
plt.ylabel("Spending Score")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

df = pd.read_csv("Market_Basket_Optimisation.csv", header=None)
transactions = [
    [str(item).strip() for item in row if pd.notna(item) and str(item).strip()]
    for row in df.values
]

encoder = TransactionEncoder()
encoded = encoder.fit(transactions).transform(transactions)
basket = pd.DataFrame(encoded, columns=encoder.columns_)
frequent = apriori(basket, min_support=0.01, use_colnames=True)

for confidence in [0.2, 0.4, 0.6]:
    rules = association_rules(frequent, metric="confidence", min_threshold=confidence)
    print("\nMinimum Confidence:", confidence)
    print(rules[["antecedents", "consequents", "support", "confidence", "lift"]].head(10))

rules = association_rules(frequent, metric="confidence", min_threshold=0.2)
if not rules.empty:
    plt.scatter(rules["support"], rules["confidence"], s=rules["lift"] * 20)
    plt.xlabel("Support")
    plt.ylabel("Confidence")
    plt.title("Association Rules")
    plt.show()
```
